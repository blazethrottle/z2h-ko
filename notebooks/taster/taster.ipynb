## T-S04 글자 짝 세기

**할 일**: 다음 셀을 실행합니다. 모든 이름에서 '어떤 글자 다음에 어떤 글자가 왔는지'를 셉니다.

**줄별 풀이**

| 줄 | 코드 | 풀이 |
|---|---|---|
| 1 | `import torch` | 숫자 계산 도구 PyTorch를 불러옵니다 |
| 2 | `chars = sorted(set(''.join(words)))` | 이름에 나오는 글자들(a~z)을 모아 정렬합니다 |
| 3 | `stoi = {s: i+1 for i, s in enumerate(chars)}` | 글자마다 1~26 번호를 붙입니다 |
| 4 | `stoi['.'] = 0` | '.'은 이름의 시작과 끝 표시이고 번호는 0입니다 |
| 5 | `itos = {i: s for s, i in stoi.items()}` | 번호를 다시 글자로 바꾸는 사전을 만듭니다 |
| 6 | `N = torch.zeros((27, 27))` | 27 × 27 빈 표를 만듭니다(앞 글자 × 다음 글자) |
| 7 | `for w in words:` | 이름 하나씩 |
| 8 | `····cs = ['.'] + list(w) + ['.']` | 앞뒤에 시작, 끝 표시를 붙입니다. emma는 . e m m a . 가 됩니다 |
| 9 | `····for a, b in zip(cs, cs[1:]):` | 이웃한 두 글자씩 짝지어 |
| 10 | `········N[stoi[a], stoi[b]] += 1` | 그 짝의 칸에 1을 더합니다 |
| 11 | `print(N[0, stoi['a']].item())` | 'a'로 시작하는 이름 수(시작 표시 다음이 a인 횟수)를 보여 줍니다 |

**보여야 하는 것**: `4410.0`

**깨달음**: 언어 모델의 출발점은 '무엇 다음에 무엇이 자주 오나'를 세는 것입니다.

```python
import torch
chars = sorted(set(''.join(words)))
stoi = {s: i+1 for i, s in enumerate(chars)}
stoi['.'] = 0
itos = {i: s for s, i in stoi.items()}
N = torch.zeros((27, 27))
for w in words:
    cs = ['.'] + list(w) + ['.']
    for a, b in zip(cs, cs[1:]):
        N[stoi[a], stoi[b]] += 1
print(N[0, stoi['a']].item())
```

## T-S05 세기만으로 이름 짓기

**할 일**: 다음 셀을 실행합니다. 센 표를 확률로 바꾸고, 그 확률대로 글자를 하나씩 뽑아 이름 10개를 만듭니다.

**줄별 풀이**

| 줄 | 코드 | 풀이 |
|---|---|---|
| 1 | `P = (N + 1) / (N + 1).sum(1, keepdim=True)` | 세기 표를 줄마다 합이 1인 확률표로 바꿉니다(+1은 한 번도 안 나온 짝을 0으로 두지 않는 장치) |
| 2 | `g = torch.Generator().manual_seed(42)` | 무작위의 출발점을 42로 고정해, 처음부터 실행하면 늘 같은 결과가 나오게 합니다 |
| 3 | `for _ in range(10):` | 이름 10개를 만듭니다 |
| 4 | `····ix = 0` | 시작 표시(0번)에서 출발합니다 |
| 5 | `····out = []` | 지은 글자를 모을 빈 목록 |
| 6 | `····while True:` | 끝 표시가 나올 때까지 반복합니다 |
| 7 | `········ix = torch.multinomial(P[ix], 1, generator=g).item()` | 지금 글자 줄의 확률대로 다음 글자 하나를 뽑습니다 |
| 8 | `········if ix == 0:` | 끝 표시가 나오면 |
| 9 | `············break` | 멈춥니다 |
| 10 | `········out.append(itos[ix])` | 뽑은 글자를 붙입니다 |
| 11 | `····print(''.join(out))` | 지어낸 이름을 보여 줍니다 |

**보여야 하는 것**: 이상하지만 이름 비슷한 단어 10개. 처음부터 순서대로 실행했다면 첫 줄은 `anugeenvi`입니다.

**깨달음**: 세기만 해도 컴퓨터가 데이터에 없던 새 이름을 지어냅니다. 다만 아직 엉성합니다.

```python
P = (N + 1) / (N + 1).sum(1, keepdim=True)
g = torch.Generator().manual_seed(42)
for _ in range(10):
    ix = 0
    out = []
    while True:
        ix = torch.multinomial(P[ix], 1, generator=g).item()
        if ix == 0:
            break
        out.append(itos[ix])
    print(''.join(out))
```

## T-S06 작은 신경망 학습

**할 일**: 다음 셀을 실행합니다. 같은 일을 이번에는 '세기' 대신 신경망이 스스로 숫자를 고쳐 가며 배웁니다. 10초쯤 걸립니다.

**줄별 풀이**

| 줄 | 코드 | 풀이 |
|---|---|---|
| 1 | `import torch.nn.functional as F` | 신경망 계산 함수 모음을 F라는 이름으로 불러옵니다 |
| 2 | `xs, ys = [], []` | 앞 글자 번호들과 다음 글자 번호들을 모을 빈 목록 |
| 3 | `for w in words:` | 이름 하나씩 |
| 4 | `····cs = ['.'] + list(w) + ['.']` | 앞뒤 표시를 붙입니다 |
| 5 | `····for a, b in zip(cs, cs[1:]):` | 이웃한 두 글자씩 |
| 6 | `········xs.append(stoi[a])` | 앞 글자 번호를 저장합니다(문제) |
| 7 | `········ys.append(stoi[b])` | 다음 글자 번호를 저장합니다(정답) |
| 8 | `xs = torch.tensor(xs)` | 목록을 텐서(숫자 상자)로 바꿉니다 |
| 9 | `ys = torch.tensor(ys)` | 정답도 텐서로 바꿉니다 |
| 10 | `W = torch.randn((27, 27), generator=g, requires_grad=True)` | 무작위 숫자로 시작하는 27 × 27 가중치. requires_grad는 '기울기를 구해 달라'는 표시입니다 |
| 11 | `for step in range(100):` | 100번 학습합니다 |
| 12 | `····logits = W[xs]` | 앞 글자마다 다음 글자 점수 한 줄씩 꺼냅니다 |
| 13 | `····loss = F.cross_entropy(logits, ys)` | 정답을 얼마나 틀렸는지(손실) 계산합니다 |
| 14 | `····W.grad = None` | 지난번 기울기를 비웁니다 |
| 15 | `····loss.backward()` | 손실을 키우는 방향(기울기)을 계산합니다 |
| 16 | `····W.data += -50 * W.grad` | 그 반대 방향으로 조금 움직입니다(50은 걸음 크기) |
| 17 | `····if step % 10 == 0 or step == 99:` | 10번마다 한 번, 그리고 마지막에 |
| 18 | `········print(step, round(loss.item(), 3))` | 학습 횟수와 손실을 보여 줍니다 |

**보여야 하는 것**: 11줄이 보이고 손실이 약 3.84에서 약 2.47로 내려갑니다.

**깨달음**: 컴퓨터가 스스로 숫자를 고쳐 틀린 정도를 줄입니다. 이것이 '학습'입니다.

```python
import torch.nn.functional as F
xs, ys = [], []
for w in words:
    cs = ['.'] + list(w) + ['.']
    for a, b in zip(cs, cs[1:]):
        xs.append(stoi[a])
        ys.append(stoi[b])
xs = torch.tensor(xs)
ys = torch.tensor(ys)
W = torch.randn((27, 27), generator=g, requires_grad=True)
for step in range(100):
    logits = W[xs]
    loss = F.cross_entropy(logits, ys)
    W.grad = None
    loss.backward()
    W.data += -50 * W.grad
    if step % 10 == 0 or step == 99:
        print(step, round(loss.item(), 3))
```

## T-S07 신경망으로 이름 짓기

**할 일**: 다음 셀을 실행합니다. 학습한 신경망의 점수를 확률로 바꿔 이름 10개를 만듭니다. 마음에 드는 결과가 나오면 공유 카드를 만들 수 있습니다.

**줄별 풀이**

| 줄 | 코드 | 풀이 |
|---|---|---|
| 1 | `for _ in range(10):` | 이름 10개 |
| 2 | `····ix = 0` | 시작 표시에서 출발합니다 |
| 3 | `····out = []` | 지은 글자를 모을 목록 |
| 4 | `····while True:` | 끝 표시가 나올 때까지 |
| 5 | `········p = F.softmax(W[ix], dim=0)` | 신경망 점수를 합이 1인 확률로 바꿉니다 |
| 6 | `········ix = torch.multinomial(p, 1, generator=g).item()` | 그 확률대로 다음 글자를 뽑습니다 |
| 7 | `········if ix == 0:` | 끝 표시면 |
| 8 | `············break` | 멈춥니다 |
| 9 | `········out.append(itos[ix])` | 글자를 붙입니다 |
| 10 | `····print(''.join(out))` | 신경망이 지은 이름을 보여 줍니다 |

**보여야 하는 것**: 또 다른 이름 10개. 처음부터 순서대로 실행했다면 첫 줄은 `gsli`입니다.

**깨달음**: 세기 방식과 신경망은 같은 데이터에서 비슷한 수준의 답에 도달합니다. 신경망은 이 방법을 훨씬 큰 문제로 넓힐 수 있습니다.

```python
for _ in range(10):
    ix = 0
    out = []
    while True:
        p = F.softmax(W[ix], dim=0)
        ix = torch.multinomial(p, 1, generator=g).item()
        if ix == 0:
            break
        out.append(itos[ix])
    print(''.join(out))
```
